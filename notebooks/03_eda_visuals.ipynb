```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sqlite3

con = sqlite3.connect('../database/bank_churn.db')
df = pd.read_sql('SELECT * FROM bank_customers', con)

sns.set(style="whitegrid")

# Boxplot for Balance (Key Insight)
# Shows churn tendencies according to user's balance
plt.figure(figsize=(10, 6))
sns.boxplot(x='churn_flag', y='balance', data=df, palette='Set2')
plt.title('Balance Distribution: Retained vs. Churned')
plt.xlabel('Churn Status (0 = Retained, 1 = Churned)')
plt.ylabel('Balance ($)')
plt.show()


# Barplot for Complaints (NumComplaints)
# Shows the correlation between the number of complaints and churn rate
plt.figure(figsize=(8, 5))
sns.barplot(x='numcomplaints', y='churn_flag', data=df, palette='Reds')
plt.title('Churn Rate by Number of Complaints')
plt.xlabel('Number of Complaints')
plt.ylabel('Churn Rate')
plt.show()


# Barplot for Number of Products
# Shows the correlation between the number of products and churn rate
plt.figure(figsize=(8, 5))
sns.barplot(x='numofproducts', y='churn_flag', data=df, palette='Blues')
plt.title('Churn Rate by Number of Products Held')
plt.xlabel('Number of Products')
plt.ylabel('Churn Rate')
plt.show()
```
